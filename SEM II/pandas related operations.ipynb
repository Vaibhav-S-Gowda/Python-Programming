# Adding a New Row


```python
new_row = {
    "rank": "AsstProf",
    "discipline": "A",
    "yrs.science.phd": 8,
    "yrs.service": 5,
    "sex": "Female",
    "salary": 85000,
}
```

```python
# Add new row using loc
df.loc[len(df)] = new_row
```

```python
df.tail()
```

```python
### Adding a new column
df["salary_in_lakhs"] = df["salary"] / 100000
```

# Filtering

```python
# Filter employees with salary greater than 120000 (Handling numericsl column)
high_salary = df[df["salary"] > 120000]
high_salary
```

```python
prof = df[df["rank"] == "Prof"]
prof
```

```python
# Multiple conditions
filtered_data = df[(df["rank"] == "Prof") & (df["salary"] > 130000)]
filtered_data
```
