```python
import pandas as pd
df = pd.read_csv('Social-Network.csv')
df.head()
```

### Check for missing values

```python
df.isna().sum()
```

### Treating Categorical Variable

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])
df.head()
```

### Scaling Numerical Variable

```python
from sklearn.preprocessing import MinMaxScaler
mm = MinMaxScaler()
df['EstimatedSalary'] = mm.fit_transform(df[['EstimatedSalary']])
df['Age'] = mm.fit_transform(df[['Age']])
df.head()
```

### Identify X and Y

```python
x = df.iloc[:,1:4]
y = df.iloc[:,4]
```

### Split into Train and Test

```python
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size = 0.2, random_state = 1, shuffle = True)
```

# Building Model

### Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
LR = LogisticRegression()
LR.fit(xtrain, ytrain)
print(LR.coef_, LR.intercept_)
print(LR.classes_)
```

### Evaluate Model

```python
ypred = LR.predict(xtest)
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

print(confusion_matrix(ytest, ypred))
```

```python
print("Accuracy: ", accuracy_score(ytest, ypred))
print("Precision: ", precision_score(ytest, ypred))
print("Recall: ", recall_score(ytest, ypred))
print("F1_score: ", f1_score(ytest, ypred))
print("roc_auc_score: ", roc_auc_score(ytest, ypred))
```
